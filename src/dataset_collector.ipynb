# Custom Eye Dataset Collector
**Controls:**
- `SPACE` — toggle auto-save (saves every 5 frames)
- `S` — manual save single frame
- `Q` — quit / move to next cell

**Per person:** straight, left, right, up, down → 30 frames each = 150 per label per person  
**Last person** goes to `val` split

```python
import cv2
import os

DATASET_DIR  = "/home/rudhratej/multimodal_driver_fatigue_monitor/src/custom_eye_dataset"

for split in ['train', 'val']:
    for label in ['open', 'closed']:
        os.makedirs(f"{DATASET_DIR}/{split}/{label}", exist_ok=True)

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

def collect(person_name, label, split='train', target=30):
    save_dir = f"{DATASET_DIR}/{split}/{label}"
    existing = len(os.listdir(save_dir))
    count    = 0
    auto     = False
    auto_ctr = 0

    cap = cv2.VideoCapture(0)
    print(f"Collecting '{label}' for {person_name} | target={target} | existing={existing}")

    while count < target:
        ret, frame = cap.read()
        if not ret:
            break

        gray  = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.1, 3, minSize=(60, 60))

        display  = frame.copy()
        eye_crop = None

        for (x, y, w, h) in faces[:1]:
            ey1 = int(h * 0.28); ey2 = int(h * 0.52)
            ex1 = int(w * 0.06); ex2 = int(w * 0.94)
            ax1 = max(x+ex1, 0);  ay1 = max(y+ey1, 0)
            ax2 = min(x+ex2, frame.shape[1])
            ay2 = min(y+ey2, frame.shape[0])
            eye_crop = frame[ay1:ay2, ax1:ax2]
            color = (0,255,0) if label=='open' else (0,0,255)
            cv2.rectangle(display, (ax1,ay1), (ax2,ay2), color, 2)
            cv2.rectangle(display, (x,y), (x+w,y+h), (255,200,0), 1)

        mode = "AUTO" if auto else "MANUAL"
        cv2.putText(display, f"{person_name} | {label.upper()} | {count}/{target} | {mode} | SPC=auto S=save Q=quit",
                    (8, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,255,0), 2)
        cv2.imshow("Collector", display)
        key = cv2.waitKey(1) & 0xFF

        if auto and eye_crop is not None and eye_crop.size > 0:
            auto_ctr += 1
            if auto_ctr % 5 == 0:
                cv2.imwrite(f"{save_dir}/{existing+count:05d}_{person_name}.jpg", eye_crop)
                count += 1

        if key == ord('s') and eye_crop is not None and eye_crop.size > 0:
            cv2.imwrite(f"{save_dir}/{existing+count:05d}_{person_name}.jpg", eye_crop)
            count += 1
        elif key == ord(' '):
            auto = not auto
            print(f"  Auto-save: {'ON' if auto else 'OFF'}")
        elif key == ord('q'):
            print(f"  Quit early at {count}")
            break

    cap.release()
    cv2.destroyAllWindows()
    print(f"Done — {count} saved for {person_name} ({label})")

print("Ready")
```
